```python
import numpy as np
```

```python
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
```

```python
data = load_breast_cancer(as_frame = True)
df = data.frame
print(df.shape)
df.describe()
df.head()
```

```python
X = data.data
Y = data.target

print(X.shape)
print(Y.shape)

w = np.zeros(30)


scaler = StandardScaler()

X = scaler.fit_transform(X)
print("X mean after scaling" , X.mean(axis = 0))
print("X std after scaling" , X.std(axis = 0))
```

```python
def sigmoid(z):
  return 1/(1+np.exp(-z))
```

```python
def predict(x , w, b):
  z= np.dot(x,w)+b
  return sigmoid(z)
```

```python
def compute_cost(x , y ,w ,b):
  m=len(x)
  y_pred = predict(x , w,b)
  return (-1/m)*np.sum(y*np.log(y_pred)+ (1-y)*np.log(1-y_pred))
```

```python
def compute_gradient(x,y,w,b):
  m = len(x)
  y_pred = predict(x,w,b)
  error = y_pred - y
  dw = (1/m)*(np.dot(x.T, error))
  db = (1/m)*(np.sum(error))
  return dw,db
```

```python
#loop
w = np.zeros(X.shape[1])
b=0
epochs = 1000
alpha = 0.001
for i in range(epochs):
  dw , db = compute_gradient(X,Y,w,b)
  w = w - alpha*dw
  b = b - alpha*db
  if i%100 == 0:
    print(compute_cost(X,Y,w,b) )
```

```python
prob = predict(X,w,b)

n=0
for i in range(len(prob)):
  if prob[i] >= 0.5:
    prob[i] = 1
  else:
    prob[i] = 0

  if prob[i] == Y[i]:
    n+=1

accuracy = n/len(prob)
print(accuracy*100)
```
